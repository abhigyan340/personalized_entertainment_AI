```python
def genre_overlap_score(movie):
    matches = movies[movies['title'].str.lower() == movie.lower()]
    
    if matches.empty:
        return None
    
    movie_index = matches.index[0]
    movie_genres = set(movies.iloc[movie_index]['genre_list'])
    
    distances = similarity[movie_index]
    
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    relevant = 0
    
    for i in movie_list:
        recommended_genres = set(movies.iloc[i[0]]['genre_list'])
        
        if movie_genres & recommended_genres:
            relevant += 1
    
    return relevant / 5
```

```python
def genre_coverage_score(movie):
    matches = movies[movies['title'].str.lower() == movie.lower()]
    
    if matches.empty:
        return None
    
    movie_index = matches.index[0]
    movie_genres = set(movies.iloc[movie_index]['genre_list'])
    
    distances = similarity[movie_index]
    
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    total_overlap = 0
    
    for i in movie_list:
        recommended_genres = set(movies.iloc[i[0]]['genre_list'])
        
        overlap = len(movie_genres & recommended_genres) / len(movie_genres)
        
        total_overlap += overlap
    
    return total_overlap / 5
# 1.0 → recommendations cover all original genres
# 0.5 → on average, half the original genres are represented
# 0.0 → no genre overlap
```

```python
def feature_coverage_score_semantic(movie):
    matches = movies[movies['title_x'].str.lower() == movie.lower()]
    
    if matches.empty:
        return None
    
    movie_index = matches.index[0]
    
    # Features of the original movie
    movie_genres = set(movies.iloc[movie_index]['genre_list'])
    movie_keywords = set(movies.iloc[movie_index]['keywords'])
    movie_cast = set(movies.iloc[movie_index]['cast'])
    movie_director = movies.iloc[movie_index]['director']
    
    # Find top 5 recommendations
    distances = embedding_similarity[movie_index]
    
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    total_score = 0
    
    for i in movie_list:
        recommended = movies.iloc[i[0]]
        
        recommended_genres = set(recommended['genre_list'])
        recommended_keywords = set(recommended['keywords'])
        recommended_cast = set(recommended['cast'])
        recommended_director = recommended['director']
        
        # Genre overlap
        genre_score = (
            len(movie_genres & recommended_genres) / len(movie_genres)
            if movie_genres else 0
        )
        
        # Keyword overlap
        keyword_score = (
            len(movie_keywords & recommended_keywords) / len(movie_keywords)
            if movie_keywords else 0
        )
        
        # Cast overlap
        cast_score = (
            len(movie_cast & recommended_cast) / len(movie_cast)
            if movie_cast else 0
        )
        
        # Director match
        director_score = (
            1 if movie_director == recommended_director
            and movie_director != ''
            else 0
        )
        
        # Combine the four signals
        recommendation_score = (
            genre_score +
            keyword_score +
            cast_score +
            director_score
        ) / 4
        
        total_score += recommendation_score
    
    return total_score / 5
```

```python
import pandas as pd
import ast
import numpy as np

movies_df = pd.read_csv("../data/tmdb_5000_movies.csv")
credits_df = pd.read_csv("../data/tmdb_5000_credits.csv")
```

```python
def extract_names(text):
    try:
        return [item["name"].lower() for item in ast.literal_eval(text)]
    except:
        return []
```

```python
movies["genre_list"] = movies["genres"].apply(extract_names)
```

```python
def extract_cast(text):
    try:
        return [
            item["name"].lower()
            for item in ast.literal_eval(text)[:5]
        ]
    except:
        return []
    
```

```python
movies["cast_list"] = movies["cast"].apply(extract_cast)
```

```python
def extract_director(text):
    try:
        crew = ast.literal_eval(text)
        
        for item in crew:
            if item["job"] == "Director":
                return item["name"].lower()
        
        return ""
    except:
        return ""
```

```python
tfidf_min = similarity.min(axis=1, keepdims=True)
tfidf_max = similarity.max(axis=1, keepdims=True)

tfidf_normalized = (
    similarity - tfidf_min
) / (tfidf_max - tfidf_min)
```

```python
import numpy as np

np.random.seed(42)

test_indices = np.random.choice(
    len(movies),
    size=100,
    replace=False
)

len(test_indices)
```

```python
def get_hybrid_similarity(tfidf_weight):
    
    semantic_weight = 1 - tfidf_weight
    
    return (
        tfidf_weight * tfidf_normalized
        + semantic_weight * semantic_normalized
    )
```

```python
def evaluate_hybrid(tfidf_weight):
    
    hybrid = get_hybrid_similarity(tfidf_weight)
    
    scores = []
    
    for index in test_indices:
        
        movie_genres = set(movies.iloc[index]['genre_list'])
        movie_keywords = set(movies.iloc[index]['keyword_list'])
        movie_cast = set(movies.iloc[index]['cast_list'])
        movie_director = movies.iloc[index]['director']
        
        if len(movie_genres) == 0:
            continue
        
        distances = hybrid[index]
        
        movie_list = sorted(
            list(enumerate(distances)),
            reverse=True,
            key=lambda x: x[1]
        )[1:6]
        
        total_score = 0
        
        for i in movie_list:
            
            recommended = movies.iloc[i[0]]
            
            recommended_genres = set(
                recommended['genre_list']
            )
            
            recommended_keywords = set(
                recommended['keyword_list']
            )
            
            recommended_cast = set(
                recommended['cast_list']
            )
            
            recommended_director = recommended['director']
            
            genre_score = (
                len(movie_genres & recommended_genres)
                / len(movie_genres)
            )
            
            keyword_score = (
                len(movie_keywords & recommended_keywords)
                / len(movie_keywords)
                if movie_keywords else 0
            )
            
            cast_score = (
                len(movie_cast & recommended_cast)
                / len(movie_cast)
                if movie_cast else 0
            )
            
            director_score = (
                1
                if movie_director
                and movie_director == recommended_director
                else 0
            )
            
            feature_score = (
                genre_score
                + keyword_score
                + cast_score
                + director_score
            ) / 4
            
            total_score += feature_score
        
        scores.append(total_score / 5)
    
    return np.mean(scores), len(scores)
```

```python
weights = [
    0.0,
    0.2,
    0.4,
    0.5,
    0.6,
    0.8,
    1.0
]

results = []

for weight in weights:
    
    score, count = evaluate_hybrid(weight)
    
    results.append({
        'tfidf_weight': weight,
        'semantic_weight': 1 - weight,
        'score': score,
        'movies_evaluated': count
    })

results
```

```python
def genre_similarity(movie_index, candidate_index):
    
    movie_genres = set(
        movies.iloc[movie_index]['genre_list']
    )
    
    candidate_genres = set(
        movies.iloc[candidate_index]['genre_list']
    )
    
    if not movie_genres:
        return 0
    
    return len(
        movie_genres & candidate_genres
    ) / len(movie_genres)
```

```python
def keyword_similarity(movie_index, candidate_index):
    
    movie_keywords = set(
        movies.iloc[movie_index]['keyword_list']
    )
    
    candidate_keywords = set(
        movies.iloc[candidate_index]['keyword_list']
    )
    
    if not movie_keywords:
        return 0
    
    return len(
        movie_keywords & candidate_keywords
    ) / len(movie_keywords)
```

```python
keyword_similarity(
    avatar_index,
    serenity_index
)
```

```python
def cast_similarity(movie_index, candidate_index):
    
    movie_cast = set(
        movies.iloc[movie_index]['cast_list']
    )
    
    candidate_cast = set(
        movies.iloc[candidate_index]['cast_list']
    )
    
    if not movie_cast:
        return 0
    
    return len(
        movie_cast & candidate_cast
    ) / len(movie_cast)
```

```python
cast_similarity(
    avatar_index,
    serenity_index
)
```

```python
def director_similarity(movie_index, candidate_index):
    
    movie_director = movies.iloc[movie_index]['director']
    candidate_director = movies.iloc[candidate_index]['director']
    
    if not movie_director or not candidate_director:
        return 0
    
    return float(
        movie_director == candidate_director
    )
```

```python
director_similarity(
    avatar_index,
    serenity_index
)
```

```python
def metadata_similarity(
    movie_index,
    candidate_index,
    genre_weight=0.40,
    keyword_weight=0.20,
    cast_weight=0.20,
    director_weight=0.20
):
    
    genre_score = genre_similarity(
        movie_index,
        candidate_index
    )
    
    keyword_score = keyword_similarity(
        movie_index,
        candidate_index
    )
    
    cast_score = cast_similarity(
        movie_index,
        candidate_index
    )
    
    director_score = director_similarity(
        movie_index,
        candidate_index
    )
    
    return (
        genre_weight * genre_score
        + keyword_weight * keyword_score
        + cast_weight * cast_score
        + director_weight * director_score
    )
```

```python
def v5_score(movie_index, candidate_index, metadata_weight=0.10):
    
    hybrid_score = hybrid_similarity[movie_index][candidate_index]
    
    metadata_score = metadata_similarity(
        movie_index,
        candidate_index
    )
    
    final_score = (
        (1 - metadata_weight) * hybrid_score
        + metadata_weight * metadata_score
    )
    
    return final_score
```

```python
def evaluate_v5_on_indices(metadata_weight, indices):
    
    scores = []
    
    for index in indices:
        
        movie_genres = set(
            movies.iloc[index]['genre_list']
        )
        
        if len(movie_genres) == 0:
            continue
        
        distances = hybrid_similarity[index]
        
        movie_list = sorted(
            list(enumerate(distances)),
            reverse=True,
            key=lambda x: x[1]
        )[1:21]
        
        reranked = []
        
        for candidate_index, hybrid_score in movie_list:
            
            metadata_score = metadata_similarity(
                index,
                candidate_index
            )
            
            final_score = (
                (1 - metadata_weight) * hybrid_score
                + metadata_weight * metadata_score
            )
            
            reranked.append(
                (candidate_index, final_score)
            )
        
        reranked.sort(
            key=lambda x: x[1],
            reverse=True
        )
        
        top_5 = reranked[:5]
        
        total_score = 0
        
        for candidate_index, _ in top_5:
            
            recommended_genres = set(
                movies.iloc[candidate_index]['genre_list']
            )
            
            genre_score = (
                len(movie_genres & recommended_genres)
                / len(movie_genres)
            )
            
            total_score += genre_score
        
        scores.append(total_score / 5)
    
    return np.mean(scores), len(scores)
```

```python
metadata_weights = [
    0.0,
    0.05,
    0.10,
    0.15,
    0.20,
    0.30,
    0.40,
    0.50
]

v5_results = []

for weight in metadata_weights:
    
    score, count = evaluate_v5_on_indices(
        weight,
        test_indices
    )
    
    v5_results.append({
        "metadata_weight": weight,
        "hybrid_weight": 1 - weight,
        "score": score,
        "movies_evaluated": count
    })

v5_results
```

```python
def evaluate_recommendations(indices, metadata_weight):
    
    semantic_scores = []
    genre_scores = []
    
    for index in indices:
        
        movie_genres = set(
            movies.iloc[index]['genre_list']
        )
        
        if len(movie_genres) == 0:
            continue
        
        # Get top 20 using the original hybrid model
        distances = hybrid_similarity[index]
        
        candidates = sorted(
            list(enumerate(distances)),
            reverse=True,
            key=lambda x: x[1]
        )[1:21]
        
        # Re-rank using V5
        reranked = []
        
        for candidate_index, hybrid_score in candidates:
            
            metadata_score = metadata_similarity(
                index,
                candidate_index
            )
            
            final_score = (
                (1 - metadata_weight) * hybrid_score
                + metadata_weight * metadata_score
            )
            
            reranked.append(
                (candidate_index, final_score)
            )
        
        reranked.sort(
            key=lambda x: x[1],
            reverse=True
        )
        
        top_5 = reranked[:5]
        
        # Evaluate the final recommendations
        semantic_total = 0
        genre_total = 0
        
        for candidate_index, _ in top_5:
            
            # Semantic relevance
            semantic_total += semantic_similarity[
                index
            ][candidate_index]
            
            # Genre relevance
            candidate_genres = set(
                movies.iloc[candidate_index]['genre_list']
            )
            
            genre_overlap = (
                len(movie_genres & candidate_genres)
                / len(movie_genres)
            )
            
            genre_total += genre_overlap
        
        semantic_scores.append(
            semantic_total / 5
        )
        
        genre_scores.append(
            genre_total / 5
        )
    
    return {
        "semantic_score": np.mean(semantic_scores),
        "genre_score": np.mean(genre_scores),
        "movies_evaluated": len(semantic_scores)
    }
```
